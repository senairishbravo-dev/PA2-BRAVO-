```python
#A. REPRODUCIBLE NORMALIZATION PROBLEM
```

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
X
```

```python
x = np.mean(X)
x
```

```python
o = np.std(X)
o
```

```python
Z = (X-x)/o
Z
```

```python
X_normalized = Z
X_normalized
```

```python
X
```

```python
xn = np.mean(Z)
xn
```

```python
xstd = np.std(Z)
xstd
```

```python
np.save('X_normalized.npy', X_normalized)
```

```python
#B. CUBES DIVISIBLE BY 4
```

```python
n = np.arange(1,101)
cubed = n ** 3
C = cubed.reshape(10,10)
C
```

```python
print("Shape of C: ",C.shape)
```

```python
div_by_4 = C[C % 4 == 0]
div_by_4
```

```python
print("number of selected elements:",div_by_4.size)
```

```python
np.save('div_by_4.npy',div_by_4)
```

```python
#c ABOVE-MEAN SQUARES PROBLEM
```

```python
n = np.arange(1,37)
n
```

```python
s = n ** 2
s
```

```python
S = s.reshape(6,6)
S
```

```python
S_mean = np.mean(S)
S_mean
```

```python
above_mean = S[S > S_mean]
above_mean
```

```python
print('Number of selected elements:', above_mean.size)
```

```python
np.save('above_mean.npy', above_mean)
```
